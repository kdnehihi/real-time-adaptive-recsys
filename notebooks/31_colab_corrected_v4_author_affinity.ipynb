# Notebook 31 — Corrected V4 + Author-Affinity Calibration

This notebook does **not retrain V4**. It loads the accepted training checkpoint from Notebook 30 and tests whether a lightweight, point-in-time author-affinity score can recover retrieval quality.

The user-disjoint protocol is:

1. tune author/popularity weights on calibration users only;
2. freeze the selected weights;
3. evaluate on different holdout users with exactly the same A2 queries, candidates, eligibility and seen-item masks;
4. compare against Corrected V4 with matched popularity and the existing Legacy V4.2;
5. promote only when paired-bootstrap HR@1000 improves and cold-item HR@1000 does not regress.

Use an **A100 or L4 GPU runtime**. A100 is faster; L4 should fit. Reports are small and are saved to Drive and downloaded locally.

## Download verified runtime and artifacts

Add a read token named `HF_TOKEN` to Colab Secrets and enable notebook access. No write token is required. Downloads are cached inside the current Colab runtime, so rerunning a cell does not upload anything.

```python
from huggingface_hub import hf_hub_download, snapshot_download

HF_REPO_ID = 'khoa1505/recsys-artifacts'
RUNTIME_FILE = 'runtime/corrected_v4_author_affinity_runtime_v1.zip'
RUNTIME_SHA256 = '588fa5a7ad6c81bd174d5c46fed8fe5c2f264422b460ef47c7c61af8b97cdf5e'
CORRECTED_V4_FILE = 'v4_corrected/v1/best_model.pt'
LEGACY_V4_FILE = 'v4/v4_inference_fp16.pt'
LEGACY_V4_SHA256 = '4ad49b3e6f9f8b5d45a6c92f31d8d9b27f558e1746613efb44a23db3cd456a26'

def normalize_token(raw):
    if isinstance(raw, bytes): raw = raw.decode()
    if isinstance(raw, str):
        raw = raw.strip()
        return raw if raw.startswith('hf_') else None
    if isinstance(raw, dict):
        for value in raw.values():
            token = normalize_token(value)
            if token: return token
    return None

try:
    HF_TOKEN = normalize_token(userdata.get('HF_TOKEN'))
except Exception:
    HF_TOKEN = None
HF_TOKEN = HF_TOKEN or normalize_token(getpass('Paste Hugging Face READ token: '))
assert HF_TOKEN, 'HF_TOKEN is missing or invalid.'

def sha256(path):
    digest = hashlib.sha256()
    with Path(path).open('rb') as stream:
        while block := stream.read(8 * 1024 * 1024): digest.update(block)
    return digest.hexdigest()

runtime_zip = Path(hf_hub_download(HF_REPO_ID, RUNTIME_FILE, repo_type='model', token=HF_TOKEN))
corrected_v4 = Path(hf_hub_download(HF_REPO_ID, CORRECTED_V4_FILE, repo_type='model', token=HF_TOKEN))
legacy_v4 = Path(hf_hub_download(HF_REPO_ID, LEGACY_V4_FILE, repo_type='model', token=HF_TOKEN))
assert sha256(runtime_zip) == RUNTIME_SHA256, 'Runtime ZIP hash mismatch'
assert sha256(legacy_v4) == LEGACY_V4_SHA256, 'Legacy V4 checkpoint hash mismatch'
assert corrected_v4.stat().st_size > 100_000_000, 'Corrected V4 checkpoint is unexpectedly small'
snapshot_download(
    HF_REPO_ID, repo_type='model', token=HF_TOKEN, local_dir=HF_SNAPSHOT_ROOT,
    allow_patterns=[
        'corrected_sequences/v2/manifest.json',
        'corrected_sequences/v2/sequences/split=validation/**',
    ],
)
corrected_manifest = json.loads((CORRECTED_ROOT / 'manifest.json').read_text())
assert corrected_manifest['quality']['passed'] is True
for source, destination in ((corrected_v4, CORRECTED_V4_DIR / 'best_model.pt'), (legacy_v4, LEGACY_V4_DIR / 'best_model.pt')):
    destination.unlink(missing_ok=True)
    destination.symlink_to(source.resolve())
print('runtime/checkpoints/corrected validation delta: PASS')
```

```python
if SOURCE_ROOT.exists(): shutil.rmtree(SOURCE_ROOT)
SOURCE_ROOT.mkdir(parents=True)
with zipfile.ZipFile(runtime_zip) as archive:
    assert archive.testzip() is None, 'Runtime archive is corrupt'
    archive.extractall(SOURCE_ROOT)
EVAL_SCRIPT = SOURCE_ROOT / 'scripts/evaluate_corrected_v4_author_affinity.py'
assert EVAL_SCRIPT.exists(), EVAL_SCRIPT
compile(EVAL_SCRIPT.read_text(), str(EVAL_SCRIPT), 'exec')
print('runtime extraction/compile: PASS')
```

## Calibrate and evaluate

This is an exact matrix-search evaluation over the full validation candidate catalog. It can take time, but it does not train either checkpoint. Keep the Colab tab connected until the final report cell completes.

```python
REPORT_DIR.mkdir(parents=True, exist_ok=True)
command = [
    sys.executable, str(EVAL_SCRIPT),
    '--adaptive-root', str(ADAPTIVE_ROOT),
    '--base-gold-root', str(BASE_GOLD_ROOT),
    '--corrected-root', str(CORRECTED_ROOT),
    '--corrected-v4-dir', str(CORRECTED_V4_DIR),
    '--legacy-v4-dir', str(LEGACY_V4_DIR),
    '--protocol-dir', str(PROTOCOL_DIR),
    '--output-dir', str(REPORT_DIR),
    '--author-weights', '0,0.05,0.1,0.15,0.2,0.25,0.3,0.4',
    '--popularity-weights', '0,0.01,0.02,0.03',
    '--legacy-author-weight', '0.2', '--legacy-popularity-weight', '0.02',
    '--calibration-user-fraction', '0.35', '--recency-tau', '3.0',
    '--query-batch-size', '16', '--item-batch-size', '8192',
    '--bootstrap-samples', '2000', '--seed', '44', '--device', 'cuda',
]
eval_log = Path('/content/corrected_v4_author_affinity.log')
process = subprocess.Popen(command, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)
with eval_log.open('w') as log:
    for line in process.stdout:
        print(line, end=''); log.write(line); log.flush()
code = process.wait()
if code != 0:
    tail = '\n'.join(eval_log.read_text(errors='replace').splitlines()[-240:])
    raise RuntimeError(f'Notebook 31 evaluation failed with exit code {code}:\n{tail}')
REPORT_JSON = REPORT_DIR / 'corrected_v4_author_affinity_evaluation.json'
assert REPORT_JSON.exists(), REPORT_JSON
print('Notebook 31 evaluation: PASS')
```

```python
import pandas as pd
report = json.loads(REPORT_JSON.read_text())
print('selected weights:', report['selected'])
print('acceptance gates:', json.dumps(report['acceptance_gates'], indent=2))
display(pd.read_csv(REPORT_DIR / 'holdout_summary.csv'))
display(pd.read_csv(REPORT_DIR / 'holdout_segments.csv').query("segment_type == 'target_warm'"))
for comparison, values in report['paired_user_bootstrap'].items():
    print('\n', comparison)
    print('HR@1000:', values['hitrate@1000'])
```

```python
# Persist only compact reporting artifacts; per-query Parquet remains local.
small_report = WORK_ROOT / 'small_report'
shutil.rmtree(small_report, ignore_errors=True); small_report.mkdir()
for name in ('corrected_v4_author_affinity_evaluation.json', 'calibration_grid.csv', 'holdout_summary.csv', 'holdout_segments.csv'):
    source = REPORT_DIR / name
    assert source.exists(), source
    shutil.copy2(source, small_report / name)
shutil.copy2(eval_log, small_report / eval_log.name)
for path in small_report.iterdir():
    shutil.copy2(path, DRIVE_REPORT_DIR / path.name)
archive = Path(shutil.make_archive('/content/v4_corrected_author_affinity_v1_report', 'zip', small_report))
print('Drive report:', DRIVE_REPORT_DIR)
print('download:', archive, f'{archive.stat().st_size / 1024**2:.2f} MiB')
files.download(str(archive))
print('FINAL DECISION:', 'ACCEPT' if report['accepted'] else 'REJECT')
```

## How to interpret the decision

- `author_affinity_gate`: author affinity beats Corrected V4 with the **same popularity weight** and its paired HR@1000 confidence interval is strictly positive.
- `legacy_promotion_gate`: the selected Corrected V4 + affinity stack beats the existing Legacy V4.2 on the same holdout users.
- `cold_item_non_regression_gate`: the branch does not trade away cold-item retrieval for warm-author gains.

If any gate fails, keep the result as diagnostic evidence and do not promote it. The next model change should then target content/cold-item representation rather than increasing the scalar affinity weight.
